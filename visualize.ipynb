```python
#animation
import matplotlib.pyplot as plt
import numpy as np
import glob
from itertools import product
import os
import matplotlib.animation as animation
import pandas as pd
current_directory = os.getcwd()
%matplotlib widget   
noise="0.50"
angle="45"
trial=0


path="/data/data_vicsek_scalar"
detail=pd.read_csv(current_directory+path+f'/Angle_{angle}/Noise_{noise}/parameters.csv') 
Lx = float(detail["Lx"])   # Size of the grid
Ly = float(detail["Lx"]) 
maxiter=int(detail["maxiter"])
Number_of_agents=int(detail["N"])
v0 = float(detail["v0"].iloc[0])
times=[]
for i in range( maxiter):
    if(i<10):tf=1
    if(i>10):tf=10
    if(i>100):tf=50
    if(i>1000):tf=100
    if(i%tf==0):times.append(i)


# Set up the figure and axis
fig, ax = plt.subplots()
ax.set_aspect('equal')
quiver = ax.quiver(0, 0, 0, 0, angles='xy', scale_units='xy', scale=1)



# Function to load data from files
def load_data(path):
    data = []
    with open(path, 'r') as f:
            for line in f:
                data.append(float(line.strip()))
    return np.array(data)
colour=['r','b','g','m']




def update(t):
    global quiver
    quiver.remove()  # Remove the previous quiver 

    data = pd.read_csv(current_directory+path+f'/Angle_{angle}/Noise_{noise}/config_data/trial_{trial}/config_{times[t]}.csv')

    
    # Load position and angle data
    px = data["x"].to_numpy()
    py = data["y"].to_numpy()
    vy = data["vy"].to_numpy()
    vx=data["vx"].to_numpy()
    
    
    # Plot the quiver plot return quiver

    ax.set_xlim(0,Lx)
    ax.set_ylim(0,Ly)
    ax.set_title("N="+str(Number_of_agents)+r"|$\eta=$"+str(noise)+r"|$\alpha=$"+str(angle)+r"|$t=$"+str(times[t]))
    quiver=ax.quiver(px, py, vx, vy,angles='xy', scale_units='xy', scale=v0)
    
    return quiver




    


# Create the animation
ani = animation.FuncAnimation(fig, update, frames=len(times), blit=False)

# Show the animation
plt.show()
```

```python
#orderparameter
import matplotlib.pyplot as plt
import numpy as np
import glob
from itertools import product
import os
import matplotlib.animation as animation
import pandas as pd
current_directory = os.getcwd()

%matplotlib widget  
plt.close('all') 
 
noise="0.50"
angle="45"
trial="0"
time="1000"
path="/data/data_vicsek_scalar"
detail=pd.read_csv(current_directory+path+f'/Angle_{angle}/Noise_{noise}/parameters.csv') 
Lx = float(detail["Lx"])   # Size of the grid
Ly = float(detail["Lx"]) 
maxiter=int(detail["maxiter"])
Number_of_agents=int(detail["N"])
v0 = float(detail["v0"].iloc[0])
numberoftrials = int(detail["numberoftrials"].iloc[0])
times=[]
for i in range( maxiter):
    if(i<10):tf=1
    if(i>10):tf=10
    if(i>100):tf=50
    if(i>1000):tf=100
    if(i%tf==0):times.append(i)


fig,ax=plt.subplots(1)

order=np.zeros(len(times))
time=np.zeros(len(times))
for t in range(numberoftrials):
    data = pd.read_csv(current_directory+path+f'/Angle_{angle}/Noise_{noise}/order_data/order_parameter_{t}_.csv')
    order+=data["va"].to_numpy()
    time=data["t"].to_numpy()
order/=numberoftrials
plt.plot(time,order)

# Show the animation
plt.show()
```
